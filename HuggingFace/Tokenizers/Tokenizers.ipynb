```python
from transformers import AutoTokenizer
```

```python
tokenizerr = AutoTokenizer.from_pretrained("sshleifer/distilbart-cnn-12-6")
```

```python
raw_inputs =[
    "I love deep learning!",
    "I hate this so much!"
]
```

```python
inputs = tokenizerr(raw_inputs, padding=True, truncation=True, return_tensors="pt")
print(inputs)
```

```python
#attention mask tell you which inputs to pay attention to, because the transformer expect each input to be same size so sometimes tokenizer adds some padding to make sure those are same size
```

```python
tokens = tokenizerr.tokenize("I live deep learning!")
```

```python
tokens
```

```python
token_ids = tokenizerr.convert_tokens_to_ids(tokens)
```

```python
token_ids
```

```python
decode_tokens = tokenizerr.decode(token_ids)
```
