```python
from sentence_transformers import SentenceTransformer, util
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
sentences = ['the cat sits outside', 'the new moview is awesome', 'the new movie is really great', 'the dog bark on stangers']
```

```python
embeddings = model.encode(sentences=sentences, convert_to_tensor=True)
```

```python
for sent,embed in zip(sentences, embeddings):
  print("Sentence:", sent)
  print("Len(Embeddings:", len(embed))
  # print("Embeddings: ", embed)
```

```python
cosine_scores = util.cos_sim(embeddings, embeddings)
```

```python
cosine_scores
```

```python
paraphrases = util.paraphrase_mining(model, sentences)
```

```python
for sim in paraphrases[0:10]:
  score, i, j = sim
  print(sentences[i], "<>" ,sentences[j], " --> ",score)
```

```python
model2 = SentenceTransformer('multi-qa-MiniLM-L6-cos-v1')
```

```python
import requests
```

```python
response = requests.get('https://raw.githubusercontent.com/laxmimerit/machine-learning-dataset/master/text-dataset-for-machine-learning/sbert-corpus.txt')
corpus = response.text.split('\r\n')

response = requests.get('https://raw.githubusercontent.com/laxmimerit/machine-learning-dataset/master/text-dataset-for-machine-learning/sbert-queries.txt')
queries = response.text.split('\r\n')
```

```python
print(corpus)
```

```python
print(queries)
```

```python
corpus_embeddings = model2.encode(corpus, convert_to_tensor=True)
queries_embeddings = model2.encode(queries, convert_to_tensor=True)
```

```python
corpus_embeddings = util.normalize_embeddings(corpus_embeddings)
queries_embeddings = util.normalize_embeddings(queries_embeddings)
```

```python
len(corpus_embeddings[0])
```

```python
hits = util.semantic_search(queries_embeddings, corpus_embeddings,score_function = util.dot_score,top_k=3)
```

```python
hits
```

```python
corpus
```

```python
for query, hit in zip(queries,hits):
    for q_hit in hit:
        id= q_hit['corpus_id']
        score = q_hit['score']

        print(query ,"<>", corpus[id],"-->",score)
    print()
```

```python
from sklearn.cluster import KMeans
```

```python
model3 = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
import requests
response = requests.get('https://raw.githubusercontent.com/laxmimerit/machine-learning-dataset/master/text-dataset-for-machine-learning/sbert-corpus.txt')
corpus = response.text.split('\r\n')
```

```python
len(corpus)
```

```python
corpus_embeddings = model3.encode(corpus)
```

```python
number_cluster = 5
```

```python
clus_model = KMeans(n_clusters=number_cluster)
clus_model.fit(corpus_embeddings)
clus_assignment = clus_model.labels_
```

```python
clus_assignment
```

```python
clustered_sent = [[] for i in range( number_cluster)]
clustered_sent
```

```python
for sen_id , clust_id in enumerate(clus_assignment):
    clustered_sent[clust_id].append(corpus[sen_id])
```

```python
for i,cluster in enumerate(clustered_sent):
    print("cluster",i+1)
    print(cluster)
    print()
```
